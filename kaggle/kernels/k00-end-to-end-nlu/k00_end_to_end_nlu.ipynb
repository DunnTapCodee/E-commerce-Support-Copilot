# End-to-end bilingual NLU
Clones branch `hung`; translates Bitext before splitting; trains and evaluates Vietnamese/English models. Outputs are saved under `/kaggle/working/module1-output`.


```python
import subprocess, sys, pathlib
print(subprocess.run(["nvidia-smi"],capture_output=True,text=True).stdout)
repo = pathlib.Path("/kaggle/working/E-commerce-Support-Copilot")
if not repo.exists():
    subprocess.run(["git", "clone", "--depth", "1", "--branch", "hung", "https://github.com/DunnTapCodee/E-commerce-Support-Copilot.git", str(repo)], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "--quiet", "-r", str(repo / "kaggle/requirements-end-to-end.txt")], check=True)
print("Repository:", repo)
```
